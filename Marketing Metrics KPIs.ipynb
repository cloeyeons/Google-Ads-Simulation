## **Setup**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

camp = pd.read_csv('/kaggle/input/analyze-the-marketing-spending/Marketing.csv')
camp['campaign_name'] = camp['campaign_name'].str.lower()
print(camp.info())
```

**Marketing KPIs**

```python
def kpis(df, group_by_col):
    report = df.groupby(group_by_col).agg({
        'impressions': 'sum', 
        'clicks': 'sum', 
        'mark_spent': 'sum',
        'leads': 'sum', 
        'orders': 'sum',
        'revenue': 'sum',
    })

    report['CTR (%)'] = (report['clicks'] / report['impressions'] * 100).round(2)
    report['CPC (₹)'] = (report['mark_spent'] / report['clicks']).round(2)
    report['CPL (₹)'] = (report['mark_spent'] / report['leads']).round(2)
    report['CAC (₹)'] = (report['mark_spent'] / report['orders']).round(2)
    report['AOV (₹)'] = (report['revenue'] / report['orders']).round(2)
    report['ROAS (%)'] = (report['revenue'] / report['mark_spent'] * 100).round(2)
    report['ROMI (%)'] = ((report['revenue'] - report['mark_spent']) / report['mark_spent'] * 100).round(2)

    kpi_cols = ['CTR (%)', 'CPC (₹)', 'CPL (₹)', 'CAC (₹)', 'AOV (₹)', 'ROAS (%)', 'ROMI (%)']
    
    return report[kpi_cols]
```

## **Performance Overview**

```python
def num_format(num):
    if abs(num) >= 1_000_000:
        return f'{num / 1_000_000:.2f}M'
    return f'{num / 1_000:.2f}K'
    
perf_overall = {
    'Metric': ['Total Impressions', 
               'Total Ad Spend (₹)', 
               'Total Orders', 
               'Total Revenue (₹)'], 
    'Value': [camp['impressions'].sum(),
              camp['mark_spent'].sum(),
              camp['orders'].sum(),
              camp['revenue'].sum()]
}

overall_table = pd.DataFrame(perf_overall)
overall_table['Value'] = overall_table['Value'].apply(num_format)
overall_table = overall_table.set_index('Metric')

overall_table
```

**Share of Ad Spend and Revenue by Campaign Category**

```python
spent_cat = camp.groupby('category')['mark_spent'].sum()
rev_cat = camp.groupby('category')['revenue'].sum()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 8))

ax1.pie(
    spent_cat, 
    labels=spent_cat.index, 
    autopct='%1.1f%%', 
)
ax1.set_title('Ad Expenditure by Category', fontsize = 20)

ax2.pie(
    rev_cat, 
    labels=rev_cat.index, 
    autopct='%1.1f%%', 
)
ax2.set_title('Revenue by Category', fontsize = 20)

plt.tight_layout()
plt.show()
```

**Daily Ad Expenditure and Revenue**

```python
daily_perf = camp.groupby('c_date')[['mark_spent', 'revenue']].sum() / 1000000

daily_perf.plot(color=['red', 'blue'], linewidth = 2, figsize = (15, 8))

plt.title("Daily Expenditure vs Revenue", fontsize = 20)
plt.ylabel("Amount (Millions ₹)")
plt.xlabel("Date")
plt.legend(["Expenditure", "Revenue"])
plt.ticklabel_format(style = 'plain', axis = 'y')
plt.tight_layout()
plt.show()
```

**CTR and ROMI Comparison**

```python
ctr_avg = date_met['CTR (%)'].mean()
romi_avg = date_met['ROMI (%)'].mean()

date_met['CTR vs Avg (%)'] = (((date_met['CTR (%)'] - ctr_avg)  / ctr_avg) * 100).round(2)
date_met['ROMI vs Avg (%)'] = (((date_met['ROMI (%)'] - romi_avg)  / romi_avg) * 100).round(2)

date_met['CTR vs Prev (%)'] = (date_met['CTR (%)'].pct_change() * 100).round(2)
date_met['ROMI vs Prev (%)'] = (date_met['ROMI (%)'].pct_change() * 100).round(2)

cols = ['CTR vs Avg (%)','ROMI vs Avg (%)','CTR vs Prev (%)', 'ROMI vs Prev (%)']
date_met_comp = date_met[cols]
date_met_comp
```

**Daily Insight**

```python
def daily_insight(date_str, date_met_comp):
    date_met_comp.index = pd.to_datetime(date_met_comp.index)

    date = pd.to_datetime(date_str)
    
    if date not in date_met_comp.index:
        return f"{date_str} is not included in the campaign period."
         
    row = date_met_comp.loc[date]

    if pd.isna(row['CTR vs Prev (%)']):
        return f"{date_str} is the first day of the campaign."
    else:
        text = f"   Daily Insight for {date_str}\n"
        text += (
            f"* CTR {'increased' if row['CTR vs Prev (%)'] > 0 else 'decreased'} "
            f"by {abs(row['CTR vs Prev (%)']):.2f}% compared to yesterday.\n"
            f"* ROMI {'increased' if row['ROMI vs Prev (%)'] > 0 else 'decreased'} "
            f"by {abs(row['ROMI vs Prev (%)']):.2f}% compared to yesterday.\n"
            f"* CTR is {abs(row['CTR vs Avg (%)']):.2f}% "
            f"{'above' if row['CTR vs Avg (%)'] > 0 else 'below'} the average.\n"
            f"* ROMI is {abs(row['ROMI vs Avg (%)']):.2f}% "
            f"{'above' if row['ROMI vs Avg (%)'] > 0 else 'below'} the average.\n"
        )
    return text
```

```python
date_1 = '2021-02-01'
print(daily_insight(date_1, date_met_comp))
```
